## Himmelblau's Function
The Himmelblau function is defined as: $f(x,y)=(x^2 +y−11)^2 +(x+y^2 −7)^2$ and has four minima at:
- -3.779310,-3.283186
- -2.805118, 3.131312
- 3.584428,-1.848126
- 3.0, 2.0

The objective is to define this function, i.e. a Python function that takes two inputs and outputs a single number.

````python
def himmel(x, y):
    return (x**2 + y - 11)**2 + (x + y**2 - 7)**2


print(round(himmel(-3.779310,-3.283186), 8))
print(round(himmel(-2.805118,3.131312), 8))
print(round(himmel(3.584428,-1.848126), 8))
print(round(himmel(3.0, 2.0), 8))

x = np.arange(-1,1)
y = np.arange(-1,1)
X, Y = np.meshgrid(x,y)
Z = himmel(X,Y)
print(X)
print(Y)
print(Z)

plt.contour(X, Y, Z)
plt.plot()
````
